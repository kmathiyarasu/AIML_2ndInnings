```python
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()
```

```python
import pandas as pd
# Fetch the dataset directly as a Pandas DataFrame
california = fetch_california_housing(as_frame=True)

# This gives you a combined DataFrame of features and targets
df = california.frame

# View the first few rows
print(df.head())
```

```python
# X will be a DataFrame containing all 8 feature columns
X = california.data

# y will be a Series containing the target column (MedHouseVal)
y = california.target
```

```python
print(X.head())
```

```python
print(y)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Plot the geographic distribution of blocks
plt.figure(figsize=(10, 7))
sns.scatterplot(
    data=X,
    x="Longitude",
    y="Latitude",
    size=X["Population"]/100,  # Scale down for clearer visual density
    alpha=0.4,
    palette="viridis"
)
plt.title("Geographic Distribution of California Housing Data Blocks")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.legend(title="Population (Scaled)")
plt.show()
```

```python
df= X.copy()
df['target'] = y
df.head()
```

```python
sns.heatmap(df.corr(),annot=True,cmap="coolwarm")
plt.title("Feature correlation Heatmap")
plt.show()
```
